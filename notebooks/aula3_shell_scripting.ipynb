## 3. Condicional if/else

```python
%%bash
ARQUIVO="/tmp/teste.fasta"
touch "$ARQUIVO"
if [ -f "$ARQUIVO" ]; then
    echo "Arquivo existe: $ARQUIVO"
else
    echo "Arquivo não encontrado"
fi
```

## 4. Loop for — processar múltiplos arquivos

```python
%%bash
# Criar arquivos de exemplo
for i in 1 2 3; do
    printf ">seq${i}\nATGCGCATTTACGC" > /tmp/amostra_${i}.fasta
done

# Processar em lote
for f in /tmp/amostra_*.fasta; do
    N=$(grep -c '>' "$f")
    echo "$(basename $f): $N sequência(s)"
done
```

## 5. Funções

```python
%%bash
contar_seqs() {
    local arquivo=$1
    grep -c '>' "$arquivo"
}

for f in /tmp/amostra_*.fasta; do
    n=$(contar_seqs "$f")
    echo "$(basename $f): $n seq"
done
```

## Exercício
Crie um script que recebe um diretório como argumento, conta as sequências em cada `.fasta` e salva um relatório em `relatorio.txt`.

```python
%%bash
cat > /tmp/relatorio.sh << 'SCRIPT'
#!/bin/bash
DIR=${1:-/tmp}
SAIDA="relatorio.txt"
echo "=== Relatório FASTA ===" > "$SAIDA"
for f in "$DIR"/*.fasta; do
    [ -f "$f" ] || continue
    N=$(grep -c '>' "$f")
    printf '%-30s %4d seqs\n' "$(basename $f)" "$N" >> "$SAIDA"
done
cat "$SAIDA"
SCRIPT
chmod +x /tmp/relatorio.sh
/tmp/relatorio.sh /tmp
```
