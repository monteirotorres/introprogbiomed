## 4. Controle de fluxo

```python
genes = ["BRCA1", "TP53", "EGFR"]
for gene in genes:
    if gene.startswith("B"):
        print(f"{gene} começa com B")
    else:
        print(f"{gene} — outro")

tamanhos = [len(g) for g in genes]  # list comprehension
print(tamanhos)
```

## 5. Funções aplicadas a sequências biológicas

```python
def gc_content(seq):
    seq = seq.upper()
    gc = seq.count('G') + seq.count('C')
    return round(gc / len(seq) * 100, 2)

def rev_comp(seq):
    tab = str.maketrans("ATGCN", "TACGN")
    return seq.upper().translate(tab)[::-1]

def analisar(seq, nome="seq"):
    return {
        "nome":     nome,
        "tamanho":  len(seq),
        "gc_pct":   gc_content(seq),
        "rev_comp": rev_comp(seq)
    }

dna = "ATGCGCATTTACGCAATG"
resultado = analisar(dna, "exemplo")
for k, v in resultado.items():
    print(f"{k}: {v}")
```

## Exercício
Escreva uma função `traduzir(seq)` que converte DNA → proteína (use apenas Met, Stop, e mais 4 aminoácidos de sua escolha).

```python
# Sua solução aqui
def traduzir(seq):
    cod = {
        'ATG': 'M', 'TAA': '*', 'TAG': '*', 'TGA': '*',
        'GCT': 'A', 'GCC': 'A', 'GCA': 'A', 'GCG': 'A',
        'TGT': 'C', 'TGC': 'C',
        'GAT': 'D', 'GAC': 'D',
        'TTT': 'F', 'TTC': 'F',
    }
    seq = seq.upper()
    prot = ''
    for i in range(0, len(seq)-2, 3):
        aa = cod.get(seq[i:i+3], '?')
        if aa == '*': break
        prot += aa
    return prot

print(traduzir("ATGGCTTGCGATTTTAA"))  # MACDF
```
