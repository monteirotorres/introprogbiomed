## 4. Controle de fluxo

```r
genes <- c("BRCA1", "TP53", "EGFR")
for (gene in genes) {
  if (startsWith(gene, "B")) {
    cat(gene, "começa com B\n")
  } else {
    cat(gene, "— outro\n")
  }
}

tamanhos <- sapply(genes, nchar)  # equivalente a list comprehension
print(tamanhos)
```

## 5. Funções aplicadas a sequências biológicas

```r
gc_content <- function(seq) {
  bases <- strsplit(toupper(seq), "")[[1]]
  gc    <- sum(bases %in% c("G","C"))
  round(gc / length(bases) * 100, 2)
}

rev_comp <- function(seq) {
  bases <- strsplit(toupper(seq), "")[[1]]
  tab   <- c(A="T",T="A",G="C",C="G",N="N")
  paste(rev(tab[bases]), collapse="")
}

analisar <- function(seq, nome="seq") {
  list(
    nome     = nome,
    tamanho  = nchar(seq),
    gc_pct   = gc_content(seq),
    rev_comp = rev_comp(seq)
  )
}

dna <- "ATGCGCATTTACGCAATG"
resultado <- analisar(dna, "exemplo")
for (k in names(resultado))
  cat(k, ":", as.character(resultado[[k]]), "\n")
```

## Exercício
Escreva uma função `conteudo_gc_multiplo(seqs)` que recebe um vetor de sequências e retorna um data.frame com colunas `seq` e `gc_pct`.

```r
conteudo_gc_multiplo <- function(seqs) {
  gc_vals <- sapply(seqs, gc_content)
  data.frame(seq=seqs, gc_pct=gc_vals, row.names=NULL)
}

seqs <- c("ATGCGC", "TTACGG", "GGCATTTACGCAATG")
print(conteudo_gc_multiplo(seqs))
```
