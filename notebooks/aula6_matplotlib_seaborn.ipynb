```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

sns.set_theme(style="whitegrid", palette="Set2")
plt.rcParams.update({"figure.dpi":120, "axes.titlesize":13})

np.random.seed(42)
n = 80
df = pd.DataFrame({
    "gene":      [f"GENE{i:03d}" for i in range(n)],
    "organismo": np.random.choice(["Homo sapiens","Mus musculus"], n),
    "condicao":  np.random.choice(["controle","tratamento"], n),
    "expressao": np.round(np.random.lognormal(1.5, 0.8, n), 3),
    "gc_pct":    np.round(np.random.uniform(35, 70, n), 1),
})
df["log2_expr"] = np.log2(df["expressao"])
df.head()
```

## 1. Scatter plot

```python
fig, ax = plt.subplots(figsize=(7, 5))
sns.scatterplot(data=df, x="gc_pct", y="log2_expr",
                hue="organismo", style="condicao",
                alpha=0.75, s=70, ax=ax)
ax.set_title("GC% vs. Expressão (log2)")
ax.set_xlabel("Conteúdo GC (%)")
ax.set_ylabel("Expressão (log2)")
plt.tight_layout()
plt.show()
```

## 2. Histograma + KDE

```python
fig, ax = plt.subplots(figsize=(7, 4))
sns.histplot(data=df, x="log2_expr", hue="condicao",
             bins=20, kde=True, ax=ax)
ax.set_title("Distribuição de Expressão")
ax.set_xlabel("Expressão (log2)")
plt.tight_layout()
plt.show()
```

## 3. Boxplot com pontos individuais

```python
fig, ax = plt.subplots(figsize=(7, 5))
sns.boxplot(data=df, x="organismo", y="log2_expr",
            hue="condicao", palette="Set2", ax=ax)
sns.stripplot(data=df, x="organismo", y="log2_expr",
              hue="condicao", dodge=True, color=".3",
              alpha=0.4, size=3, ax=ax,
              legend=False)
ax.set_title("Expressão por Organismo e Condição")
plt.tight_layout()
plt.show()
```

## 4. Heatmap de correlação

```python
gene_cols = [f"GENE{i:03d}" for i in range(8)]
df_wide = df.head(20).pivot_table(
    index="condicao", columns="gene", values="expressao"
)
corr = df_wide.T.corr()

fig, ax = plt.subplots(figsize=(6, 5))
sns.heatmap(corr, annot=True, fmt=".2f",
            cmap="RdBu_r", center=0,
            vmin=-1, vmax=1, square=True,
            ax=ax)
ax.set_title("Correlação")
plt.tight_layout()
plt.show()
```

## 5. Figura combinada e exportação

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.boxplot(data=df, x="condicao", y="log2_expr",
            palette="pastel", ax=axes[0])
sns.stripplot(data=df, x="condicao", y="log2_expr",
              color=".4", alpha=0.5, size=4, ax=axes[0])
axes[0].set_title("Expressão por Condição")

sns.scatterplot(data=df, x="gc_pct", y="log2_expr",
                hue="organismo", alpha=0.7, ax=axes[1])
axes[1].set_title("GC% vs. Expressão")

plt.suptitle("Análise de Expressão Gênica", fontsize=14, y=1.02)
plt.tight_layout()
plt.savefig("/tmp/figura_final.png", dpi=300, bbox_inches="tight")
plt.show()
print("Figura salva em /tmp/figura_final.png")
```
